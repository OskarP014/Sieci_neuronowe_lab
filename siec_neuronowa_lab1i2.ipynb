```python
import numpy as np
import pandas as pd
from sklearn.datasets import load_iris


iris = load_iris()
X = iris.data
y = iris.target

X_min = np.min(X, axis=0)
X_max = np.max(X, axis=0)
X_norm = (X - X_min) / (X_max - X_min)


def licz_odleglosc(probka, wagi) :
    return np.sqrt(np.sum((probka - wagi) ** 2))


def znajdz_zwyciezce(probka, wagi_wszystkich_neuronow) :
    odleglosci = np.linalg.norm(wagi_wszystkich_neuronow - probka, axis=1)
    return np.argmin(odleglosci)


def aktualizuj_wagi(wagi, indeks, probka, alpha) :
    wagi[indeks] = wagi[indeks] + alpha * (probka - wagi[indeks])
    return wagi



def trenuj_som(X_train, epoki=50, alpha_0=0.5) :
    np.random.seed(42)
    wagi = np.random.rand(3, 4)
    T = epoki


    for t in range(epoki) :
        alpha_t = alpha_0 * (1 - (t / T))


        if t % 10 == 0 :
            print(f"Epoka: {t}, współczynnik uczenia: {alpha_t:.2f}")

        for probka in X_train :
            zwyciezca = znajdz_zwyciezce(probka, wagi)
            wagi = aktualizuj_wagi(wagi, zwyciezca, probka, alpha_t)

    print(f"Epoka: {epoki}, współczynnik uczenia: {alpha_t:.2f}")
    return wagi


wagi_koncowe = trenuj_som(X_norm, epoki=50, alpha_0=0.5)

print("\nKońcowe wagi 3 neuronów:")
print(np.round(wagi_koncowe, 4))


def ocena_sieci(X_dane, y_prawdziwe, wagi) :
    macierz = np.zeros((3, 3), dtype=int)

    for i in range(len(X_dane)) :
        probka = X_dane[i]
        klasa = y_prawdziwe[i]

        zwyciezca = znajdz_zwyciezce(probka, wagi)
        macierz[zwyciezca, klasa] += 1

    df = pd.DataFrame(
        macierz,
        columns=['Setosa', 'Versicolor', 'Virginica'],
        index=['N1', 'N2', 'N3']
    )

    df.index.name = 'Neuron'
    return df


macierz_wynikow = ocena_sieci(X_norm, y, wagi_koncowe)
print("\nMacierz przypisań")
print("\n",macierz_wynikow)
```
